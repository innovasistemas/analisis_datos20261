# **Transformacion de variables**

**Codificacion one-hot:** esta tecnica implica crear una columna binaria para cada categoria de la variable categorica. La columna tendra un valor de 1 si la categoria esta presente y 0 en caso contrario. Pandas tiene una funcion get_dummies que realiza esta codificacion.

```python
# Transformar la variable de metodo de pago
df=pd.get_dummies(df, columns=['Método de pago'])
df
```

**Codificacion ordinal:** esta tecnica implica asignar un numero unico a cada categoria de la variable en orden de importancia o frecuencia

```python
# Transformar la categoria de productos
df['Categoría de productos comprados']=df['Categoría de productos comprados'].map({'Ropa':1, 'Juguetes':2, 'Electrónica':3, 'Alimentos':4})
df
```

```python
# Renombrar las columnas
df.columns=['Id_cliente', 'Ultima_fecha_compra', 'Frecuencia_compra','Valor_promedio_compra','Categoria_compra', 'Edad', 'Pago_efectivo','Pago_credito','Pago_transferencia']
df
```
