# **Limpieza de datos**

```python
df_hurtos_MDE.isnull().sum()
```

```python
# Eliminar las columnas
df_hurtos_MDE.drop(columns=['latitud', 'longitud', 'modelo'], inplace=True)
```

```python
df_hurtos_MDE.info()
```

```python
# Ordenar de manera ascendente los nombres de los barrios
pd.unique(df_hurtos_MDE.nombre_barrio.sort_values(ascending=True))
```

```python
# Renombrar los registros
df_hurtos_MDE.nombre_barrio.replace(
     { "U.D. Atanasio Girardot":"Estadio",
       "La Palma S.C.":"La Palma",
       "Naranjal S.C.":"Naranjal",
       "San Pablo A.V.":"San Pablo",
       "Las Palmas S.E.":"Las Palmas",
       "Las Playas S.C.":"Las Playas",
       "La Florida S.A.P.":"La Florida",
       "La Esperanza A.V.":"La Esperanza",
       "Facultad de Minas U. Nal":"Facultad de Minas U. Nacional",
       "Pajarito S.C.":"Pajarito",
       "Cerro Nutibara Ins.":"Cerro Nutibara",
       "Batallón Cuarta Brigada":"Cuarta Brigada"}, inplace=True)
```

```python
# Verificamos los registros en barrio
pd.unique(df_hurtos_MDE.nombre_barrio)
```

```python
# Cambiar el tipo de datos en algunas columnas del dataset

## Cambiar tipo de dato de algunos campos
df_hurtos_MDE = df_hurtos_MDE.astype({'edad': 'uint8',
                                           'cantidad':'uint8',
                                          'estado_civil':'category',
                                          'sexo':'category',
                                          'bien':'category',
                                  'actividad_delictiva':'category',
                                  'sede_receptora':'category',
                                  'color':'category',
                                  'modalidad':'category'
                           })
df_hurtos_MDE.info()
```

```python
# Creamos una copia del df original
df_hurtos_MDE_Limp=df_hurtos_MDE
```

```python
# Agrupamos por sexo y almacenamos el resultado en un variable
df_hurtos_MDE_sexo=df_hurtos_MDE_Limp.groupby('sexo').size()
df_hurtos_MDE_sexo
```

```python
# Tratar los valores atipicos en edad
df_hurtos_MDE_Limp.edad.replace({
                           -1: None,
                           121:None,
                           255:None}, inplace =True)
```

```python
# Verificamos nuevamente los campos de edad
pd.unique(df_hurtos_MDE_Limp['edad'])
```

```python
# Identificamos cantidades por   edad
df_hurtos_MDE_edad= df_hurtos_MDE_Limp.groupby('edad').size().sort_values(ascending=True)
df_hurtos_MDE_edad
```

```python
# Identificamos las cantiades por barrio
df_hurtos_MDE_barrio=df_hurtos_MDE_Limp.groupby('nombre_barrio').size().sort_values(ascending=False)
df_hurtos_MDE_barrio
```

```python
# Identificamos cantidades por bien
df_hurtos_MDE_bien= df_hurtos_MDE_Limp.groupby('bien').size().sort_values(ascending=False)
df_hurtos_MDE_bien
```

```python
# Identificamos por estado civil
df_hurtos_MDE_estado_civil=df_hurtos_MDE_Limp.groupby('estado_civil').size().sort_values(ascending=False)
df_hurtos_MDE_estado_civil
```

```python
# Agrupar por varios campos
hurtos_barrio_sexo=df_hurtos_MDE_Limp.groupby(['sexo', 'nombre_barrio']).size().sort_values(ascending=False)
hurtos_barrio_sexo
```

```python
hurtos_edad_sexo=df_hurtos_MDE_Limp.groupby(['sexo', 'edad']).size().sort_values(ascending=False)
hurtos_edad_sexo
```

```python
df_hurtos_MDE_bien=df_hurtos_MDE_Limp.groupby(['sexo', 'nombre_barrio', 'bien']).size().sort_values(ascending=False)
df_hurtos_MDE_bien
```

```python
pd.unique(df_hurtos_MDE_Limp.modalidad)
```
